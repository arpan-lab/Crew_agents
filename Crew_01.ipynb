```python
import os
from crewai import LLM

llm = LLM(
    model="groq/openai/gpt-oss-20b",
    temperature=0.1
)

print("Groq GPT-OSS LLM configured ✅")
```

```python
#Create the agent
from crewai import Agent

researcher = Agent(
    role="Market Research Analyst",
    goal="Analyze important trends in the AI industry.",
    backstory="An expert analyst with a keen eye for market trends.",
    tools=[],
    verbose=True,
    llm=llm
)

writer = Agent(
    role="Content Writer",
    goal="Write engaging content about the AI industry.",
    backstory="A skilled technology writer.",
    tools=[],
    verbose=True,
    llm=llm
)
```

```python
from crewai import Task

research = Task(
    description="Analyze important trends in the AI industry.",
    expected_output="A summary of the top 3 AI developments and their impact.",
    agent=researcher
)

write = Task(
    description="Write an engaging blog post based on the research.",
    expected_output="A 4-paragraph markdown blog post.",
    agent=writer
)
```

```python
from crewai import Crew

crew = Crew(
    agents=[researcher, writer],
    tasks=[research, write],
    verbose=True
)
```

```python
result = await crew.kickoff_async()

print(result)
```
